# 4. Data Cleaning

Data cleaning prepares the dataset for analysis by handling missing values, correcting data types, removing duplicates, and validating data quality.

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], dayfirst=True)
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"], dayfirst=True)
```

## Handling Missing Values

Before performing analysis, we inspect the dataset for missing values. Since only the `Postal_Code` column contains a small number of missing values and it is not critical for our planned analysis, these records will be removed.


```python
# Check missing values

df.isnull().sum()
```

```python
# Remove rows with missing postal codes

df = df.dropna(subset=["Postal_Code"])
```

```python
# Verify missing values have been removed

df.isnull().sum()
```

```python
# Convert Postal_Code to integer

df["Postal_Code"] = df["Postal_Code"].astype(int)
```

```python
df.dtypes
```

## Duplicate Records

Duplicate records can lead to incorrect analysis by counting the same transaction multiple times. We check for duplicates and remove them if necessary.


```python
# Check duplicate records

print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows

df = df.drop_duplicates()
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

## Data Validation

To ensure data quality, we verify that every order was shipped on or after the order date.


```python
# Check for invalid shipping dates

invalid_dates = df[df["Ship_Date"] < df["Order_Date"]]

print("Invalid Records:", len(invalid_dates))
```

```python
# Final shape after cleaning

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Save cleaned dataset

df.to_csv("../data/cleaned/superstore_sales_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

# 5. Feature Engineering

Feature engineering involves creating new variables from existing data to improve analysis and generate meaningful business insights. These new features will help identify sales trends, seasonal patterns, and shipping performance.


```python
# Extract year from Order_Date

df["Order_Year"] = df["Order_Date"].dt.year
```

```python
df[["Order_Date", "Order_Year"]].head()
```

```python
# Extract month number

df["Order_Month"] = df["Order_Date"].dt.month
```

```python
df[["Order_Date","Order_Month"]].head()
```

```python
# Extract month name

df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df[["Order_Date","Month_Name"]].head()
```

```python
# Extract quarter

df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df[["Order_Date","Quarter"]].head()
```

```python
# Calculate shipping time in days

df["Shipping_Days"] = (df["Ship_Date"] - df["Order_Date"]).dt.days
```

```python
df[["Order_Date","Ship_Date","Shipping_Days"]].head()
```

# 6. Exploratory Data Analysis (EDA)

The objective of this section is to explore the cleaned dataset, identify trends, and answer important business questions using descriptive statistics and visualizations.

## Business Question 1

### Which region generates the highest sales revenue?


```python
# Sales by Region

region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

region_sales
```

### Business Insight

The West region generated the highest sales, indicating strong customer demand and market performance. Regions with comparatively lower sales may require additional marketing efforts or operational improvements.

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

category_sales
```

### Business Insight

Technology products contribute the highest sales, followed by Furniture and Office Supplies. The company should continue investing in high-performing categories while identifying opportunities to improve lower-performing categories.

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

segment_sales
```

### Business Insight

The Consumer segment contributes the largest share of sales, making it the company's primary customer base.

```python
monthly_sales = df.groupby("Month_Name")["Sales"].sum()

monthly_sales
```

```python
# Monthly sales trend

month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

monthly_sales = (
    df.groupby("Month_Name")["Sales"]
      .sum()
      .reindex(month_order)
)

monthly_sales
```

```python
top_customers = (
    df.groupby("Customer_Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

A relatively small number of customers contribute significantly to total sales. Retaining these high-value customers can have a meaningful impact on revenue.

```python
top_cities = (
    df.groupby("City")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_cities
```

```python
shipping = df["Ship_Mode"].value_counts()

shipping
```

```python
top_products = (
    df.groupby("Product_Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```
